```python
data = pd.read_excel('health_screenings_2020_1000ea.xlsx')
data3 = data.loc[:,['city_code', 'gender', 'waist']]
```

```python
mandata = data3[data3['gender']==1].groupby('city_code')['waist'].mean()
womandata = data3[data3['gender']==2].groupby('city_code')['waist'].mean()


plt.figure(figsize=(12, 4))

plt.plot(mandata, color='b', marker='o', linestyle='--', label='M')
plt.plot(womandata, color='r', marker='o', linestyle='--', label='W')

plt.xlim(10, 50) 
plt.xlabel('count')
plt.ylabel('waist')
plt.title('2020 health screenings Waist)
plt.legend()
plt.grid()
plt.show()
```
