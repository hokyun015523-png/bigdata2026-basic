## 조건에 맞는 데이터 추출

```python
# 최고로 더웠던 날의 모든 정보(행)
W = df['temp'] == df['temp'].max()
df[W]
```

```python
df[df['temp'] == df['temp'].max()]
```

## 조건이 2개 이상 --> 그리고 &, 또는 | (역슬러시)


```python
# 기온이 30도 이상이고, 최대풍속이 9이상인 데이터

W= (df['temp'] >=30) & (df['max_wind'] >=9)
df[W]
```

```python
# 기온이 30도 이상이거나, 최대풍속이 9이상인 데이터
W= (df['temp'] >=30) | (df['max_wind'] >=9)
df[W]
```
