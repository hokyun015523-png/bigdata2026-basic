# 산점도 그래프

```python
import pandas as pd
import matplotlib.pyplot as plt
import koreanize_matplotlib
```

```python
# 데이터 불러오기
data = pd.read_excel('health_screenings_2020_1000ea.xlsx')
```

```python
data.shape
```

```python
data.head()
```

## 키, 몸무게 데이터를 가지고 산점도 그래프 그리기

```python
heightdata = data['height']
weightdata = data['weight']

plt.figure(figsize=(10, 4))
plt.scatter(x=heightdata, y=weightdata)
plt.title('2020년 검강건진 산점도 그래프')
plt.xlabel('키')
plt.ylabel('몸무게')

plt.grid()
plt.show()
```

## 수축기혈압,이완기 혈압, 공복 혈당 그룹 산전도 그래프

```python
systolic_data = data['systolic'] # 수축기 혈압
diastolic_data = data['diastolic'] # 이완기 혈압
blood_sugar_data = data['blood_sugar'] # 공복 혈당

plt.figure(figsize=(10, 6))


plt.scatter(systolic_data, diastolic_data,  color='r', label='수축기*이완기')
plt.scatter(systolic_data, blood_sugar_data, color='g', label='수축기*공복혈당')
plt.scatter(diastolic_data,blood_sugar_data, color='b', label='이완기*공복혈당')

plt.show() #--> 너무 다 밀집되어있어서 알아보기 힘들다
```

```python
systolic_data = data['systolic'] # 수축기 혈압
diastolic_data = data['diastolic'] # 이완기 혈압
blood_sugar_data = data['blood_sugar'] # 공복 혈당

plt.figure(figsize=(10, 6))


plt.scatter(systolic_data, diastolic_data, color='r', edgecolor='w', label='수축기*이완기')
plt.scatter(systolic_data, blood_sugar_data, color='g', edgecolor='w', label='수축기*공복혈당')
plt.scatter(diastolic_data,blood_sugar_data, color='b', edgecolor='w', label='이완기*공복혈당')

plt.title('2020년 건강검진 그룹 산점도 그래프')
plt.legend() #범례
plt.grid()
plt.xlim(40, 180)
plt.ylim(30, 300)

plt.show()
```

## 여성 키 히스토그램

```python
# 여성의 데이터를 추출
data.head()
```

```python
fe = data['gender'] ==2
```

```python
female_data = data.loc[fe]
female_data.head()
```

```python
female_data = data.loc[fe,['gender', 'height']]
female_data.head
```

```python
plt.figure(figsize=(10, 6))
plt.hist(female_data['height'], bins=7, label='Female')
#키의 분포를 알고 싶다 7개 구간으로 나우어져있다
plt.title('2020년 건강검진 여성 키 히스토그램')
plt.xlim(130, 180)
plt.xlabel('키')
plt.ylabel('빈도수(frequency)')
plt.grid()
plt.show()
```

## 남성, 여성 키 그룹 히스토그램

```python
ma = data['gender'] == 1   # 남성은 1
```

```python
male_data = data.loc[ma, ['gender', 'height']]
male_data.head()
```

```python
plt.figure(figsize=(10, 6))
plt.hist(male_data['height'], bins=7, alpha= 0.5, label='남성')
# alpha =0.5--> 투명도가 50%
plt.hist(female_data['height'], bins=7, alpha= 0.5, label='여성')

plt.title('2020년 건강검진 남성, 여성 키 그룹 히스토그램')
plt.xlim(130, 195)
plt.xlabel('키')
plt.ylabel('빈도수(frequency)')

plt.legend()
plt.grid()
plt.show()
```

## 남성, 여성 키 상자수염 그래프
