#꽃데이터의 길이 시각화 - 막대그래프

```python
iris.sepal_length[:10].plot(kind='bar',rot=0) #10개의 데이터만 막대그래프,회정0도
plt.title('꽃받침의 길이 시각화')# 차트 제목
plt.xlabel('Data')#x축 제목
plt.ylabel('꽃받침의 길이')#y축 제목
plt.show()
```

## 붓꽃의 품종별 그룹화 - 평균 = 새로운 데이터프레임으로 저장 -> 차트 생성

```python
df2 = iris.groupby(iris['species']).mean()
df2
```

### 품종별 꽃받침의 길이 그래프

```python
df2.sepal_length[:].plot(kind='bar', rot=0)
plt.title('품종별 꽃받침 길이 그래프')
plt.xlabel('품종')
plt.ylabel('꽃받침의 길이의 평균값')
plt.show()
```

## 타이타닉 데이터

```python
# Sex: 성별 , Survived: 생존 --> 성별에 따른 생존율
titanic =pd.read_csv('train.csv')
```

```python
titanic
```

```python
titanic_2 = titanic[['Sex', 'Survived']] # 성별 + 생존율
titanic_2
```

```python
titanic_2.groupby('Sex').mean()
```

```python
# 함께 승선한 부모와 자식의 수(Parch)에 따른 생존율

titanic_3 = titanic[['Parch', 'Survived']]
titanic_3
```

```python
titanic_3.groupby('Parch').mean()
```

```python
titanic_3.groupby('Parch', as_index=False).mean() 
```

```python
titanic_3.groupby('Parch', as_index=False).mean().sort_values('Survived', ascemding=False) 
```

```python
### 3 객실 등급, 성별 컬럼별로 생존율 나이의 평균값


titanic_4 = titanic[['Pclass', 'Sex', 'Survived', 'Age']]
titanic_4
```

```python
titanic_4.groupby(['Pclass', 'Sex']).mean()
```

#### 데이터프레임명. groupby(['그룹할컬럼1', '그룹할 칼럼2'])[['계산하고싶은컬럼1', '계산하고싶은컬럼2']].mean()

```python
titanic_4.groupby(['Pclass', 'Sex'])[['Survived','Age']].mean()
```
