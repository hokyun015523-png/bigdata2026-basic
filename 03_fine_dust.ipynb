```python
## 2018년도 지역별 미세먼지 선 그래프 
```

```python
data = pd.read_excel('fine_dust.xlsx', index_col='area') # 인덱스 행---> area(지역명)로
data
```

```python
data2018 = data[2018]  # 1차원 배열, 시리즈 형식
data2018
```

```python
plt.figure(figsize=(15,4)) # 가로는 15인치, 세로 4인치
plt.plot(data2018, color='b', marker='o')
plt.title('2018년 미세먼지 선 그래프')
plt.xlabel('지역')
plt.ylabel('micrometer')
plt.grid() # 격자무늬
plt.show()
```

## 2016년 ~ 2019년 까지 지역별 미세먼지 선 그래프

```python
for year in range(2016,2020):
    chartdata = data[year]
    print(chartdata)
    print('-' * 50)
```

```python
plt.figure(figsize=(15, 4))
for year in range(2016,2020):
    chartdata = data[year]
    plt.plot(chartdata, marker='s', label=year)

plt.title('2016~2019 미세먼지 선 그래프')
plt.xlabel('지역')
plt.ylabel('micrometer')
plt.legend()
plt.grid() # 격자무늬
plt.show()
```

```python
data = pd.read_excel('fine_dust.xlsx', index_col='area') #->  이 버전은 인덱싱 부분을 변경
data
```

```python
data4 = data.loc['Gyeonggi':'Daegu'] 
data4
```

```python
data4 = data4.loc[:, 2016:2019]
data4
```

```python
plt.figure(figsize=(15, 4))
for year in range(2016, 2020):
    chartdata = data4[year]
    plt.bar(data4.index, chartdata, label=year) # index 범위를 안줬다.

plt.show()
#--> 이건 틀린거다
```

```python
import numpy as np
index = np.arange(4) #0~3

plt.figure(figsize=(15, 4))
for year in range(2016, 2020):
    chartdata = data4[year]
    plt.bar(index, chartdata, width=0.2, label=year) 
    index = index +0.2  #인덱스 위치를 확인

plt.show()
```

```python
import numpy as np
index = np.arange(4) #0~3

plt.figure(figsize=(15, 4))
for year in range(2016, 2020):
    chartdata = data4[year]
    plt.bar(index, chartdata, width=0.2, label=year) 
    index = index +0.2  

plt.ylim(35, 55) # 범위를 조금더 세분화 하겠다라는거다
plt.xlabel('지역')
plt.ylabel('micrometer')
plt.title('2016 ~ 2019 경기, 인천, 부산, 대구 지역별 미세먼지 그룹 세로막대')
plt.legend()
plt.show()
```
