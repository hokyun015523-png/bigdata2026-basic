## 컬럼(열, 변수, 피처) 삭제 / 생성

### drop(칼럼명, axis=1) --> 컬럼(열) 삭제

- drop(...., axis=0) --> 행 삭제 ( axis=0은 생략 가능)

```python
# 지역, 망, 측정소코드 삭제

dust.drop(['지역', '망', '측정소코드'], axis=1)
```

```python
dust.head()
```

```python
dust = dust.drop(['지역', '망', '측정소코드'], axis=1)
```
