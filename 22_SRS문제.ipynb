# 1. 게임캐릭터 만들기

```python
# game_character.py
# 이름:___ # 날짜:___

class Character:
    """게임 캐릭터 클래스"""

    def __init__(self, name, hp, attack, job='모험가', shield=0):
        self.name = name #캐릭터 이름
        self.hp = hp #현대 체력
        self.attack = attack
        self.shield = shield
        self.is_blocking = False 
```

```python
def show_info(self):
    #캐릭터 정보를 출력하세요(R-04, R-16)
    print(f"[{self.job}] {self.name} | hp: {self.hp} | 공력력: {self.attack} | 방패: {self.shield}")
```
